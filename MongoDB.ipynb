# Q1. What are the key differences between SQL and NoSQL databases?

### Answer:
| SQL | NoSQL |
|---|---|
| Relational database | Non-relational database |
| Uses tables | Uses collections/documents |
| Fixed schema | Flexible schema |
| Uses SQL language | Stores JSON-like data |

Examples:
- SQL → MySQL
- NoSQL → MongoDB

# Q4. How does MongoDB ensure high availability using replication?

### Answer:
MongoDB uses Replica Sets.

Replica Set contains:
- One Primary node
- Multiple Secondary nodes

Data is copied from primary to secondary nodes.
If the primary fails, another node automatically becomes primary.

```python
import pandas as pd
from pymongo import MongoClient

# Load CSV File
df = pd.read_csv("Superstore.csv")

# MongoDB Connection
client = MongoClient("mongodb://localhost:27017/")

# Create Database
db = client["SuperstoreDB"]

# Create Collection
collection = db["Orders"]

# Convert DataFrame to Dictionary
data = df.to_dict(orient="records")

# Insert Data
collection.insert_many(data)

print("Dataset inserted successfully.")

# Question 2: Retrieve and print all documents from the Orders collection.
Answer:
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Fetch Documents
for document in collection.find():
    print(document)

# Question 3: Count and display the total number of documents in the Orders collection.
**Answer:**
from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017/")
db = client["SuperstoreDB"]
collection = db["Orders"]

* Count Documents
count = collection.count_documents({})

print("Total Documents:", count)

# Question 4: Write a query to fetch all orders from the "West" region.
Answer:
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Query
west_orders = collection.find({"Region": "West"})

for order in west_orders:
    print(order)
    
# Question 5: Write a query to find orders where Sales is greater than 500.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Query
orders = collection.find({"Sales": {"$gt": 500}})

for order in orders:
    print(order)
    
# Question 6: Fetch the top 3 orders with the highest Profit.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Top 3 Profit Orders
top_orders = collection.find().sort("Profit", -1).limit(3)

for order in top_orders:
    print(order)
    
# Question 7: Update all orders with Ship Mode as "First Class" to "Premium Class."
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Update Query
result = collection.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)

print("Documents Updated:", result.modified_count)

# Question 8: Delete all orders where Sales is less than 50.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Delete Query
result = collection.delete_many(
    {"Sales": {"$lt": 50}}
)

print("Documents Deleted:", result.deleted_count)

# Question 9: Use aggregation to group orders by Region and calculate total sales per region.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Aggregation Pipeline
pipeline = [
    {
        "$group": {
            "_id": "$Region",
            "TotalSales": {
                "$sum": "$Sales"
            }
        }
    }
]

results = collection.aggregate(pipeline)

for result in results:
    print(result)
    
# Question 10: Fetch all distinct values for Ship Mode from the collection.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Distinct Values
ship_modes = collection.distinct("Ship Mode")

print(ship_modes)

# Question 11: Count the number of orders for each category.
**Answer:**
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")

db = client["SuperstoreDB"]

collection = db["Orders"]

# Aggregation Query
pipeline = [
    {
        "$group": {
            "_id": "$Category",
            "TotalOrders": {
                "$sum": 1
            }
        }
    }
]

results = collection.aggregate(pipeline)

for result in results:
    print(result)
